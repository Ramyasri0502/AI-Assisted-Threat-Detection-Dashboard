```python
import sqlite3

def main():
    # 1. Connect to database (creates file if not exists)
    conn = sqlite3.connect("students.db")
    cursor = conn.cursor()

    # 2. Create table
    cursor.execute("""
    CREATE TABLE IF NOT EXISTS students (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT,
        age INTEGER
    )
    """)

    # 3. Insert data
    cursor.execute("INSERT INTO students (name, age) VALUES (?, ?)", ("Alice", 20))
    cursor.execute("INSERT INTO students (name, age) VALUES (?, ?)", ("Bob", 22))
    cursor.execute("INSERT INTO students (name, age) VALUES (?, ?)", ("Charlie", 19))
    conn.commit()

    # 4. View all records
    print("\n--- All Students ---")
    cursor.execute("SELECT * FROM students")
    for row in cursor.fetchall():
        print(row)

    # 5. Filter records
    print("\n--- Students older than 20 ---")
    cursor.execute("SELECT * FROM students WHERE age > 20")
    for row in cursor.fetchall():
        print(row)

    # 6. Update a record
    cursor.execute("UPDATE students SET age = 21 WHERE name = 'Alice'")
    conn.commit()

    # 7. Delete a record
    cursor.execute("DELETE FROM students WHERE name = 'Charlie'")
    conn.commit()

    # 8. Count records
    cursor.execute("SELECT COUNT(*) FROM students")
    print("\nTotal students:", cursor.fetchone()[0])

    # 9. Order records
    print("\n--- Students ordered by age ---")
    cursor.execute("SELECT * FROM students ORDER BY age ASC")
    for row in cursor.fetchall():
        print(row)

    # 10. Close connection
    conn.close()

if __name__ == "__main__":
    main()

```

Write a Python program that connects to an SQLite database named "company.db" and performs SQL operations using pandas for display.
The program should:
- Create a table named 'students' with columns: id, name, and age.
- Insert sample student records and display the table.
- Select and display students older than 20.
- Update the age of student 'Alice' to 21.
- Delete the student named 'Charlie'.
- Display the updated 'students' table.
- Create a table named 'employees' with columns: id, name, department, and salary.
- Insert sample employee records and display the table.
- Use GROUP BY to show the total salary per department.
- Create a table named 'projects' with columns: project_id, project_name, and employee_id (linked to employees).
- Insert sample project records and display the table.
- Perform a JOIN to display employees with their assigned projects.
- Use a subquery to display employees earning above the average salary.
- Close the database connection at the end.


```python
import sqlite3
import pandas as pd

# Create Database Connection
conn = sqlite3.connect("company.db")
cursor = conn.cursor()

# --- Create Students Table ---
cursor.execute("DROP TABLE IF EXISTS students")
cursor.execute("""
CREATE TABLE students (
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER
)
""")

# Insert Records
cursor.executemany("INSERT INTO students (id, name, age) VALUES (?, ?, ?)", [
    (1, 'Alice', 20),
    (2, 'Bob', 22),
    (3, 'Charlie', 19)
])
conn.commit()

print("\nStudents Table:")
df_students = pd.read_sql_query("SELECT * FROM students", conn)
print(df_students)

# Filter Records
print("\nStudents older than 20:")
df_filter = pd.read_sql_query("SELECT * FROM students WHERE age > 20", conn)
print(df_filter)

# Update Record
cursor.execute("UPDATE students SET age = 21 WHERE name = 'Alice'")
conn.commit()

# Delete Record
cursor.execute("DELETE FROM students WHERE name = 'Charlie'")
conn.commit()

print("\nStudents after update & delete:")
print(pd.read_sql_query("SELECT * FROM students", conn))

# --- Create Employees Table ---
cursor.execute("DROP TABLE IF EXISTS employees")
cursor.execute("""
CREATE TABLE employees (
    id INTEGER PRIMARY KEY,
    name TEXT,
    department TEXT,
    salary REAL
)
""")

# Insert Employee Records
cursor.executemany("INSERT INTO employees (id, name, department, salary) VALUES (?, ?, ?, ?)", [
    (1, 'Ravi', 'HR', 35000),
    (2, 'Sneha', 'IT', 55000),
    (3, 'Arjun', 'Finance', 45000),
    (4, 'Meera', 'IT', 60000)
])
conn.commit()

print("\nEmployees Table:")
print(pd.read_sql_query("SELECT * FROM employees", conn))

# Group By
print("\nTotal salary per department:")
print(pd.read_sql_query("SELECT department, SUM(salary) AS total_salary FROM employees GROUP BY department", conn))

# --- Create Projects Table ---
cursor.execute("DROP TABLE IF EXISTS projects")
cursor.execute("""
CREATE TABLE projects (
    project_id INTEGER PRIMARY KEY,
    project_name TEXT,
    employee_id INTEGER,
    FOREIGN KEY (employee_id) REFERENCES employees(id)
)
""")

# Insert Project Records
cursor.executemany("INSERT INTO projects (project_id, project_name, employee_id) VALUES (?, ?, ?)", [
    (101, 'Website Development', 2),
    (102, 'Payroll System', 3),
    (103, 'Mobile App', 4)
])
conn.commit()

print("\nProjects Table:")
print(pd.read_sql_query("SELECT * FROM projects", conn))

# Join Employees with Projects
print("\nEmployees with their projects:")
print(pd.read_sql_query("""
SELECT e.name, e.department, p.project_name
FROM employees e
JOIN projects p ON e.id = p.employee_id
""", conn))

# Subquery
print("\nEmployees earning above average salary:")
print(pd.read_sql_query("""
SELECT * FROM employees
WHERE salary > (SELECT AVG(salary) FROM employees)
""", conn))

# Close Connection
conn.close()
```
